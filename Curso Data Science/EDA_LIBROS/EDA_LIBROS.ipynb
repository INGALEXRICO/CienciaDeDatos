```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# CARGAR EL DATASET

```python
df = pd.read_csv("libros_toscrape_pag1.csv")
print("Carga de datos exitosa!!")
```

```python
#Gráfica por Rating
```

```python
sns.countplot(x="rating", data=df)
plt.title("Distribución de libros por Rating")
plt.xlabel("Rating")
plt.ylabel("Cantidad de libros")
plt.show()
```

```python
sns.histplot(df["precio_gbp"], kde=True, bins=20)
plt.title("Distribución de precios de los libros")
plt.xlabel("Precio (GBP)")
plt.ylabel("Cantidad de libros")
plt.show()
```
