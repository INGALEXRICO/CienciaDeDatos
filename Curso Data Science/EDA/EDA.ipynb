```python
#importar las librerias que vamos a ocupar
import pandas as pd #carga de datos
import numpy as np #Analisis de datos
import matplotlib.pyplot as plt #Gráficas
```

```python
# Cargar el archivo de datos (dataframe)
df = pd.read_csv('titanic.csv')
print("carga de datos exitosa!!")
```

```python
# inspección inicial
df.head() #cargar las primeras 5 filas
```

```python
df.info() #tipos de datos que tenemos
```

```python
df.describe() #ESTADISTICAS BASICAS (MEDIA,MIN,MAX)
```

```python
df.describe(include='all') #incluye hasta las columnas que nos 
```

```python
#df.head()
#df.info()
#df.describe()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median()) ##rellenar la columna age #Rellenar la columna Age con el valor de la mediana
```

```python
df.head(10)
```

```python
df.dropna(subset=['Embarked'], inplace=True) #Eliminar filas con valores faltantes
```

```python
df.drop('Cabin', axis=1, inplace=True) #Eliminar la columna que tiene datos faltantes
```

```python
df['Pclass'] = df['Pclass'].astype('category') #Convertir la columna pclass
```

```python
df.info()
```

```python
#Configuración estética
import seaborn as sns
sns.set(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (10, 6)
```

```python
#Variable objetivo
sns.countplot(x="Survived", data=df)
plt.title("Distribución de sobrevivientes")
plt.show()
```

```python
#Variables categóricas
cat_vars = ["Sex", "Pclass", "Embarked"]
```

```python
for col in cat_vars:
    sns.countplot(x=col, data=df)
    plt.title(f"Distribución de {col}")
    plt.show()
```

```python
#Variables numéricas
num_vars = ["Age", "Fare"]
```

```python
for col in num_vars:
    sns.histplot(df[col].dropna(), kde=True, bins=30)
    plt.title(f"Distribución de {col}")
    plt.show()
```

```python
#Supervivencia por sexo
sns.barplot(x="Sex", y="Survived", data=df)
plt.title("Sobrevivientes según sexo")
plt.show()
```

```python
#Supervivencia por clase
sns.barplot(x="Pclass", y="Survived", data=df)
plt.title("Sobrevivientes según clase")
plt.show()
```

```python
#Supervivencia por edad
sns.barplot(x="Survived", y="Age", data=df)
plt.title("Sobrevivientes según edad")
plt.show()
```

```python
#Correlaciones
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Mapa de correlaciones")
plt.show()
```

```python
#Guardar el nuevo dataset limpio
df.to_csv('datos_nuevos_titanic.csv', index=False)
```
