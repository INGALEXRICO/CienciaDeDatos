```python
#funciones

def clasificar_edad(edad):
    """Recibe una edad y regresa la categoria correspondiente."""
    if edad < 12:
        return "niño"
    elif edad < 18:
        return "adolescente"
    elif edad < 65:
        return "adulto"
    else: 
        return "adulto mayor"

#probamos la funcion con distintos valores
for edad in [5,15,40,80]:
    print(edad, "->", clasificar_edad(edad))
```

```python
pasajeros = [pasajero, pasajero2]

print("CANTIDAD DE PASAJEROS:", len(pasajeros))

while len(pasajeros) < 4:

    espacios = 4 - len(pasajeros)

    print("Todavía hay espacio para", espacios, "pasajeros")

    nombre = input("Ingresa el nombre: ")
    edad = int(input("Ingresa la edad: "))
    clase = int(input("Ingresa la clase: "))
    tarifa = float(input("Ingresa la tarifa: "))

    respuesta = input("¿Sobrevivió? (si/no): ")
    sobrevivio = respuesta.lower() == "si"

    nuevo_pasajero = {
        "nombre": nombre,
        "edad": edad,
        "clase": clase,
        "tarifa": tarifa,
        "sobrevivio": sobrevivio
    }

    pasajeros.append(nuevo_pasajero)

    print("Pasajero agregado correctamente")
    print("Cantidad actual:", len(pasajeros))


print("Ya se llenó la lista")
print(*pasajeros, sep="\n")


# *pasajeros= Separa los elementos que están dentro de la lista
# sep="\n"= "Entre cada elemento, pon un salto de línea
```

```python
# Creamos una función llamada agregar_pasajero.
# Esta función recibe una lista y los datos de un pasajero.
def agregar_pasajero(lista, nombre, edad, clase, tarifa, sobrevivio):

    # Creamos un diccionario llamado pasajero.
    # Aquí guardamos todos los datos del pasajero.
    pasajero = {
        "nombre": nombre,          # Guarda el nombre
        "edad": edad,              # Guarda la edad
        "clase": clase,            # Guarda la clase en la que viajaba
        "tarifa": tarifa,          # Guarda cuánto pagó
        "sobrevivio": sobrevivio   # Guarda True si sobrevivió y False si no
    }

    # Agregamos el diccionario del pasajero a la lista.
    lista.append(pasajero)


# Creamos una lista vacía.
# Aquí se van a guardar todos los pasajeros.
lista_pasajeros = []


# El ciclo se repetirá mientras haya menos de 4 pasajeros.
while len(lista_pasajeros) < 4:

    # Preguntamos al usuario si quiere agregar un pasajero.
    opcion = input(
        "¿Desea agregar un pasajero? Presione 1 para sí y 0 para no: "
    )

    # Si el usuario escribe 1, entramos aquí.
    if opcion == "1":

        # Pedimos el nombre del pasajero.
        nombre = input("Ingresa el nombre: ")

        # Pedimos la edad y la convertimos a número entero.
        edad = int(input("Ingresa la edad: "))

        # Pedimos la clase y la convertimos a número entero.
        clase = int(input("Ingresa la clase: "))

        # Pedimos la tarifa y la convertimos a número decimal.
        tarifa = float(input("Ingresa la tarifa: "))

        # Preguntamos si el pasajero sobrevivió.
        respuesta = input("¿Sobrevivió? (si/no): ")

        # Convertimos la respuesta a minúsculas.
        # Si escribió "si", sobrevivio será True.
        # Si escribió cualquier otra cosa, será False.
        sobrevivio = respuesta.lower() == "si"

        # Llamamos a la función agregar_pasajero.
        # Le mandamos la lista y todos los datos que capturamos.
        agregar_pasajero(
            lista_pasajeros,
            nombre,
            edad,
            clase,
            tarifa,
            sobrevivio
        )

        # Mostramos un mensaje indicando que se agregó correctamente.
        print("Pasajero agregado correctamente")

        # len() nos dice cuántos pasajeros hay actualmente en la lista.
        print("Cantidad de pasajeros:", len(lista_pasajeros))

    # Si el usuario no escribió 1, entra al else.
    else:

        print("LISTA LLENA")
        # break termina el ciclo while.
        break


# Cuando termina el ciclo mostramos un título.
print("\nLISTA DE PASAJEROS:")

# Mostramos cada pasajero en una línea diferente.
# El * permite sacar los elementos de la lista.
# sep="\n" hace que cada pasajero aparezca en una nueva línea.
print(*lista_pasajeros, sep="\n")
```
